```python
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

print("Shape after removing outliers:", df.shape)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA

url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"

df = pd.read_csv(url, parse_dates=["Month"], index_col="Month")
passengers = df["Passengers"].asfreq("MS")

print(passengers.head())
```

```python
print("First five observations:")
print(passengers.head())

print("Last five observations:")
print(passengers.tail())
```

```python
print("Starting period:", passengers.index.min())
print("Ending period:", passengers.index.max())
```

```python
print("Frequency: 12 observations per year")
print("Data type:", type(passengers))
print("Number of observations:", len(passengers))
print("Time interval:", passengers.index.freq)
```

```python
plt.figure(figsize=(10, 5))
plt.plot(passengers, label="Passengers")
plt.title("Monthly Air Passengers (1949–1960)")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.legend()
plt.grid(True)
plt.show()
```

```python
decomposition = seasonal_decompose(passengers, model="multiplicative", period=12)

print("Decomposition completed")
```

```python
decomposition.plot()
plt.suptitle("AirPassengers Time Series Decomposition", y=1.02)
plt.tight_layout()
plt.show()
```

```python
plot_acf(passengers, lags=40)
plt.title("Autocorrelation Function (ACF)")
plt.show()

plot_pacf(passengers, lags=40, method="ywm")
plt.title("Partial Autocorrelation Function (PACF)")
plt.show()
```

```python
model = ARIMA(passengers, order=(1, 1, 1))
fitted_model = model.fit()

print(fitted_model)
```

```python
print(fitted_model.summary())
```

```python
forecast_result = fitted_model.get_forecast(steps=12)

forecast = forecast_result.predicted_mean
confidence_interval = forecast_result.conf_int()

forecast_df = pd.DataFrame({
    "Forecast": forecast,
    "Lower Bound": confidence_interval.iloc[:, 0],
    "Upper Bound": confidence_interval.iloc[:, 1]
})

print(forecast_df)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(passengers, label="Actual Passengers")
plt.plot(forecast, label="Forecasted Passengers", color="red")

plt.fill_between(
    confidence_interval.index,
    confidence_interval.iloc[:, 0],
    confidence_interval.iloc[:, 1],
    alpha=0.2,
    color="red",
    label="95% Confidence Interval"
)

plt.title("AirPassengers: Actual and Forecasted Values")
plt.xlabel("Year")
plt.ylabel("Number of Passengers")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df = pd.read_csv(r"C:\Users\srira\Downloads\dataset.csv(1).txt")

print("Packages loaded successfully")
display(df.head(6))
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
numeric_data = df.select_dtypes(include="number")
numeric_data = numeric_data.replace([np.inf, -np.inf], np.nan)
numeric_data = numeric_data.dropna()

scaler = StandardScaler()
data_scaled = scaler.fit_transform(numeric_data)

print("Scaled data:")
print(data_scaled[:6])
```

```python
pca = PCA()
pca_result = pca.fit_transform(data_scaled)

print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("Cumulative Variance:")
print(np.cumsum(pca.explained_variance_ratio_))

print("PCA Summary:")
print(pd.DataFrame({
    "Principal Component": [f"PC{i+1}" for i in range(len(pca.explained_variance_ratio_))],
    "Standard Deviation": np.sqrt(pca.explained_variance_),
    "Proportion of Variance": pca.explained_variance_ratio_,
    "Cumulative Proportion": np.cumsum(pca.explained_variance_ratio_)
}))
```

```python
variance_percent = pca.explained_variance_ratio_ * 100

plt.figure(figsize=(8, 5))
plt.bar(range(1, len(variance_percent) + 1), variance_percent)

for i, value in enumerate(variance_percent):
    plt.text(i + 1, value + 0.5, f"{value:.1f}%", ha="center")

plt.title("PCA Scree Plot")
plt.xlabel("Principal Component")
plt.ylabel("Percentage of Variance")
plt.ylim(0, 60)
plt.xticks(range(1, len(variance_percent) + 1))
plt.tight_layout()
plt.show()
```

```python
pca_scores = pd.DataFrame(
    pca_result[:, :2],
    columns=["PC1", "PC2"]
)

plt.figure(figsize=(10, 7))
sns.scatterplot(data=pca_scores, x="PC1", y="PC2", color="blue")

loadings = pca.components_.T[:, :2]

for i, name in enumerate(numeric_data.columns):
    plt.arrow(0, 0, loadings[i, 0] * 3, loadings[i, 1] * 3,
              color="red", alpha=0.7, head_width=0.08)
    plt.text(loadings[i, 0] * 3.2, loadings[i, 1] * 3.2,
             name, color="red", ha="center")

plt.title("PCA Biplot")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)
plt.tight_layout()
plt.show()
```
